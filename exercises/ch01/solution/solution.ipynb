### ✏ Task T1 - The short document wins

Here is the book's BM25 ranking for the query `cat dog forest` on the 12-document MINI collection,
computed with **k1 = 1.2**, **b = 0.75**, and average document length **avgdl = 9.67**. The third
column counts the occurrences of (`cat`, `dog`, `forest`) in each document:

| Rank | Doc | Length | tf (cat, dog, forest) | BM25 |
|---|---|---:|:---:|---:|
| 1 | b9 | 3 | (1, 1, 1) | 2.930 |
| 2 | b1 | 8 | (2, 2, 1) | 2.836 |
| 3 | b10 | 14 | (2, 2, 2) | 2.568 |

Document **b9** wins even though it contains the *fewest* query-term occurrences: b1 and b10 both
contain strictly more of every query term, yet rank below it.

1. Explain **which part of the BM25 formula** produces this outcome, and trace it using the
   numbers above.
2. **Predict** how the ordering of b9, b1, and b10 changes if you set **b = 0**, and justify your
   prediction. You will be able to check it against your own scorer later in the notebook.
3. You control one document and want it to rank first for `cat dog forest`, with k1 and b fixed at
   1.2 and 0.75. **How would you craft that document** to attack the scoring function, and where
   are the limits of what such an attack can gain?

<div style="background:#0F766E; color:#ffffff; padding:0.35em 0.8em; margin:0.6em 0 0.4em 0; border-radius:4px; font-weight:600; text-transform:uppercase; font-size:0.78em; letter-spacing:0.06em;">Solution</div>

**1. Document-length normalization is the cause.** The BM25 term-frequency factor is

$$\frac{\text{tf}\,(k_1+1)}{\text{tf} + k_1\,(1 - b + b\,\tfrac{|D|}{\text{avgdl}})}.$$

The denominator grows with document length through the factor $1 - b + b\,|D|/\text{avgdl}$. For
b9 ($|D|=3$, well below avgdl) this factor is $1 - 0.75 + 0.75\cdot 3/9.67 = 0.483$, so each
occurrence is divided by a small number and a single occurrence of each term already contributes
strongly (weight about 1.39 per term before IDF). For b10 ($|D|=14$) the factor is $1 - 0.75 +
0.75\cdot 14/9.67 = 1.336$, so even *two* occurrences of a term are scaled down. Term-frequency
**saturation** compounds this: the second and later occurrences of a term add progressively less.
The net effect is that b9's three well-covered terms in a short document edge out b1's and b10's
diluted repetitions.

**2. With b = 0 the ordering flips.** Setting $b=0$ removes the length factor entirely: the
denominator becomes $\text{tf} + k_1$ for every document, so length no longer penalizes anyone.
Now term frequency (after saturation) decides, and more occurrences help. b10, with two of every
query term, gets the saturated weight of tf=2 (about 1.375) on all three terms; b1 gets it on two
terms; b9 only ever has tf=1 (weight 1.0). So the prediction is **b10 > b1 > b9**: the long
document that repeats every term now wins, and the "short document wins" effect disappears.

**3. Craft a short document containing all three terms once.** BM25 sums evidence *across* the
query terms, so a winning document must contain `cat`, `dog`, and `forest` (missing a term forfeits
its whole IDF contribution). Within one term, repetition is nearly useless: saturation caps each
term's weight below $k_1+1 = 2.2$ no matter how often it repeats, so stuffing "dog dog dog ..."
buys almost nothing after the first two occurrences and only lengthens the document. Because the
length factor rewards short documents, the strongest attack is exactly b9: a minimal document that
mentions each query term once and adds no other words. In short, match every query term, keep the
document as short as possible, and do not pad. This is why BM25 resists keyword stuffing but can
still be gamed by tiny exact-match documents.

### 💻 Task C1 - Make the queries find their movies

A search engine only ever matches the *terms* it extracts, so the extraction pipeline quietly
decides what can be found. Your job is to turn a raw string into a list of terms so that a handful

```python
import tasks   # the text-processing functions live here - open the file and read them

def extract_terms(text: str) -> list[str]:
    """Turn a raw string into a list of search terms by composing functions from `tasks`."""
    # YOUR CODE HERE
    return tasks.stem(tasks.fold_case(tasks.tokenize(text)))
```

```python
# --- Given: a tiny title-only search built on your extract_terms ---
def find_movies(query: str) -> set[str]:
    """Titles whose extracted terms contain every extracted query term."""
    q = {t for t in extract_terms(query) if t}
    if not q:
        return set()
    return {title for title in movie_titles.values() if q <= set(extract_terms(title))}
```

### 💻 Task C2 - A BM25 search engine over the movies

You now have a pipeline that decides what counts as a term. Wrap it in a ranking engine. Implement
`BM25Scorer` in `tasks.py` so it indexes the movies once and answers ranked queries end to end,
from raw query string to ranked movie ids. Use the **positive (Lucene) IDF** and the BM25 formula
from Task T1.

Interface:

- `BM25Scorer(documents: dict[str, str], extract_terms, k1: float = 1.2, b: float = 0.75)` -
  `documents` maps a document id to its raw text; `extract_terms` is the term-extraction function
  you built above. The constructor extracts terms for every document and precomputes the
  document frequencies, the document count `N`, and the average document length `avgdl`.
- `idf(term: str) -> float` - positive BM25 IDF, $\log\!\big(1 + \frac{N - \text{df} + 0.5}{\text{df} + 0.5}\big)$.
- `search(query: str, k: int = 10) -> list[tuple[str, float]]` - extract the query's terms, score
  every document, keep only positive scores, and return the top `k` as `(doc_id, score)` pairs
  sorted by descending score, breaking ties by ascending id. An empty query, or one whose terms
  match nothing, returns `[]`.

```python
# --- Setup C2 ---
from tasks import BM25Scorer

scorer = BM25Scorer(movies_text, extract_terms)   # reuses the pipeline you built above

def show(query, k=5):
    display_md(f"**Top {k} for `{query}`:**")
    print_table(
        [[f"{score:.3f}", movie_titles[doc_id]] for doc_id, score in scorer.search(query, k=k)],
        headers=["BM25", "Title"],
    )

show("toy story")
```

```python
# --- Verify C2 ---
def top_title(query):
    res = scorer.search(query, k=1)
    return movie_titles[res[0][0]] if res else None

# The obvious top hit for a clear query.
assert top_title("toy story") == "Toy Story", f"got {top_title('toy story')}"
assert top_title("jurassic park dinosaurs") == "Jurassic Park", f"got {top_title('jurassic park dinosaurs')}"

# Ranking contract.
res = scorer.search("love", k=5)
assert len(res) == 5, f"k should cap results at 5, got {len(res)}"
scores = [s for _, s in res]
assert scores == sorted(scores, reverse=True), "results must be sorted by descending score"

# Edge cases.
assert scorer.search("", k=5) == [], "empty query should return []"
assert scorer.search("qzzxwv", k=5) == [], "a query matching nothing should return []"

print("✓ BM25Scorer passes all checks!")
```

### ✏ Task T2 - Scaling to a million movies

Your `search` method scores **every** document in the collection for **every** query. That is fine
for 500 movies, but imagine the collection grows to 1,000,000 movies and the engine must answer
many queries per second.

Explain why the current approach becomes too slow, and describe a better strategy that avoids
touching most documents for a typical short query. You have not studied the relevant data structure
yet (it comes in a later chapter), so reason from first principles: given a query term, what would
you want to look up instantly, and what would you precompute to make that possible?

<div style="background:#0F766E; color:#ffffff; padding:0.35em 0.8em; margin:0.6em 0 0.4em 0; border-radius:4px; font-weight:600; text-transform:uppercase; font-size:0.78em; letter-spacing:0.06em;">Solution</div>

**Why it is too slow.** There are two separate costs. A naive scorer would run the whole extraction
pipeline (tokenize, fold case, stem) over every document *on every query*, repeating the same
expensive text processing again and again. The reference implementation avoids that by design: it
extracts the terms once in `__init__` and stores them, along with the document frequencies and
lengths, so a query reuses precomputed data instead of re-tokenizing the collection. What it does
*not* avoid is the scan itself: `search` still loops over all $N$ documents and computes a score
for each, so the per-query cost is $O(N)$. At $N = 1{,}000{,}000$
that is a million score computations per query, almost all of them wasted, because a short query
like `toy story` only appears in a tiny fraction of the documents. Every document that shares *no*
query term scores exactly zero, yet we still visit it. Precomputing the tokens helped, but without
an index structure we have no way to jump to the relevant documents and must walk the entire
collection.

**A better strategy: only score documents that contain a query term.** For each *term*, precompute
the list of documents that contain it, together with the term frequency (a **postings list**).
Given a query, look up the postings list of each query term instantly, take the union of those
lists, and compute BM25 only for that small candidate set; every other document is guaranteed to
score zero and can be skipped. Precompute alongside it the per-term document frequency (for IDF)
and each document's length (for the length-normalization factor), so scoring a candidate needs no
extra passes over the text. The lookup "which documents contain this term" is exactly what an
**inverted index** provides, and it turns the per-query cost from "proportional to the whole
collection" into "proportional to how many documents actually contain the query terms". (Chapter 4
develops inverted indexes and the algorithms that walk these postings lists efficiently.)
